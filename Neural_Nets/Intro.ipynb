```python
y=data[["Exited"]]
X=data.drop(labels="Exited",axis=1)
```

```python
y.value_counts()
```

```python
cat_variables=[]
num_variables=[]
for i in X.columns:
    if X[i].dtype=="object":
        cat_variables.append(i)
    else:
        num_variables.append(i)
```

```python
num_variables
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,stratify=y,random_state=42)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler

preprocessor=ColumnTransformer([
    ("encoder" , OneHotEncoder(drop='first'),cat_variables),
    ("scaler",StandardScaler(),num_variables)
])

```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

```python
feature_names = preprocessor.get_feature_names_out()
feature_names
```

```python
X_train_df = pd.DataFrame(X_train,
    columns=feature_names
)
```

```python
X_test_df = pd.DataFrame(X_test,
    columns=feature_names
)
```

```python
X_train=X_train_df
X_test=X_test_df
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

lr=LogisticRegression()
param={
    'C':[1,10,100,100],
    'solver':['lbfgs','linear','rbf'],
    'penalty':['l1','l2'],
    'max_iter':[1000],
    'class_weight':[
        {0:0.63,2:2.46},
        {0:1,1:10},
        {0:0.3,1:0.7}
    ]
}
grid=GridSearchCV(estimator=lr,param_grid=param)
```

```python
grid.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix

y_pred=grid.predict(X_test)
acc=accuracy_score(y_test,y_pred)
acc
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
print(classification_report(y_test,y_pred))
```

```python
grid.best_estimator_
```

```python
from sklearn.tree import DecisionTreeClassifier

dt=DecisionTreeClassifier(criterion="gini")
dt.fit(X_train,y_train)
y_pred=dt.predict(X_test)
acc=accuracy_score(y_test,y_pred)
```

```python
import tensorflow as tf
```

```python
from tensorflow.keras.models import load_model,Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.losses import BinaryCrossentropy
from tensorflow.keras.metrics import Accuracy
from tensorflow.keras.optimizers import SGD
```

```python
model=Sequential([
    Dense(64,input_shape=(X_train.shape[1],), activation="relu"),
    Dense(32,activation="relu"),
   Dense (1,activation="sigmoid")
])
```

```python
model.summary()
```

```python
model.compile(optimizer="SGD",loss="BinaryCrossentropy",metrics=['accuracy'])
import datetime
```

```python
log_dir="logs/fit/"
```

```python
from tensorflow.keras.callbacks import TensorBoard

tensorboard=TensorBoard(log_dir=log_dir, histogram_freq=1)
```

```python
from tensorflow.keras.callbacks import EarlyStopping

stopping=EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
```

```python
history=model.fit(X_train,y_train,
                  validation_data=(X_test,y_test),batch_size=64,
                  epochs=100,
                  callbacks=[stopping,tensorboard])
```
